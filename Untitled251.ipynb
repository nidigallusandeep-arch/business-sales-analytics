# Sales Analysis

# 1 What is the total sales by city

```python
total_sales=df.groupby("City")["Total_Sales"].sum().sort_values(ascending=0)
total_sales
```

```python
total_sales.plot(kind="bar")
plt.title("the total sales by city")
plt.xlabel("City")
plt.ylabel("total_sales")
plt.show()
```

# 2 What is the total sales by product?

```python
total_sales=df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=0)
total_sales
```

```python
total_sales.plot(kind="bar")
plt.title("the total sales by product")
plt.xlabel("Product")
plt.ylabel("Total_sales")
plt.show()
```

# 3.What is the total sales by category?

```python
total_sales=df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=0)
total_sales
```

```python
total_sales.plot(kind="bar")
plt.title("The total sales by category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

# 4.Which city generates the highest sales?

```python
highest_sales=df.groupby("City")["Total_Sales"].max().sort_values(ascending=0)
highest_sales
```

```python
highest_sales.plot(kind="bar")
plt.title("City generates the highest sales")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.show()
```

# 6.Which product has the highest sales

```python
highest_sales=df.groupby("Product")["Total_Sales"].max().sort_values(ascending=0)
highest_sales
```
